```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["FraudFound_P"])
y = df["FraudFound_P"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
age_median = X_train.loc[X_train["Age"] != 0, "Age"].median()

print("Training median age:", age_median)
```

```python
X_train["Age"] = X_train["Age"].replace(0, age_median)
X_test["Age"] = X_test["Age"].replace(0, age_median)
```

```python
print("Age = 0 in training:", (X_train["Age"] == 0).sum())
print("Age = 0 in testing:", (X_test["Age"] == 0).sum())
```

```python
categorical_cols = X_train.select_dtypes(include="object").columns.tolist()

print("Number of categorical columns:", len(categorical_cols))
print(categorical_cols)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ],
    remainder="passthrough"
)
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)
```

```python
print("Before encoding:", X_train.shape)
print("After encoding:", X_train_encoded.shape)

print("Before encoding:", X_test.shape)
print("After encoding:", X_test_encoded.shape)
```

```python
from xgboost import XGBClassifier
```

```python
xgb_model = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=42,
    eval_metric="logloss"
)
```

```python
xgb_model.fit(X_train_encoded, y_train)
```

```python
y_pred = xgb_model.predict(X_test_encoded)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=["Genuine", "Fraud"]
))
```

```python
negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("Genuine claims:", negative)
print("Fraud claims:", positive)
print("Scale Pos Weight:", scale_pos_weight)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Genuine", "Fraud"]
)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix - XGBoost")
plt.show()
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of features:", len(feature_names))
print(feature_names[:10])
```

```python
importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": xgb_model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

importance_df.head(10)
```

```python
import matplotlib.pyplot as plt

top_features = importance_df.head(10).sort_values(
    by="Importance"
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"],
    color="steelblue"
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Feature Importances - XGBoost")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import roc_auc_score

y_prob = xgb_model.predict_proba(X_test_encoded)[:, 1]

roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", roc_auc)
```
